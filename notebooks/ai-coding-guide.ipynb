## 3. Effective Prompting Techniques

The quality of AI assistance depends heavily on how you communicate your needs.

### 3.1 The CLEAR Framework

**C** - Context: Provide relevant background
**L** - Language/Libraries: Specify tech stack
**E** - Examples: Show expected input/output
**A** - Accuracy requirements: Define constraints
**R** - Result format: Specify how you want the answer

````python
# Example: Poor Prompt
poor_prompt = """Write a function to process data"""

# Example: Good Prompt (using CLEAR)
good_prompt = """
Context: I'm building a data pipeline for user analytics

Language: Python 3.11 with pandas

Task: Write a function that:
1. Takes a DataFrame with columns: user_id, timestamp, event_type
2. Groups events by user_id
3. Calculates session duration (gap > 30 min = new session)
4. Returns summary statistics per user

Example input:
| user_id | timestamp           | event_type |
|---------|--------------------|-----------|
| u1      | 2024-01-01 10:00   | click     |
| u1      | 2024-01-01 10:15   | view      |

Expected output columns: user_id, total_sessions, avg_session_duration

Include type hints and a docstring.
"""

print("The good prompt is ~10x longer but will produce much better results")
````

### 3.2 Prompting Patterns

**1. Step-by-Step Request**
```
Help me implement user authentication:
1. First, explain the approach you'll take
2. Then implement the core logic
3. Add error handling
4. Write tests
```

**2. Constraint Specification**
```
Write a sorting algorithm with these constraints:
- Must be in-place (O(1) extra space)
- Must be stable
- No recursion (to avoid stack overflow)
```

**3. Role Assignment**
```
Act as a senior Python developer reviewing code.
Focus on: security, performance, and maintainability.
Provide specific suggestions, not just criticism.
```

**4. Iterative Refinement**
```
Start with a simple implementation, then:
- Add input validation
- Optimize for large datasets
- Make it async-compatible
```

## 4. Best Practices for AI-Assisted Development

### 4.1 The Human-AI Collaboration Model

```
+------------------+     +------------------+
|     HUMAN        |     |       AI         |
+------------------+     +------------------+
| - Requirements   |<--->| - Code generation|
| - Architecture   |     | - Boilerplate    |
| - Final review   |     | - Explanations   |
| - Testing logic  |     | - Pattern impl   |
| - Security audit |     | - Documentation  |
+------------------+     +------------------+
```

**Key Principle**: AI augments your capabilities; it doesn't replace your judgment.

### 4.2 Workflow Integration

**Recommended Workflow:**

1. **Plan** - Define what you want to build (human)
2. **Scaffold** - Let AI generate boilerplate (AI)
3. **Review** - Check generated code critically (human)
4. **Iterate** - Refine with specific feedback (human + AI)
5. **Test** - Write and run tests (AI generates, human validates)
6. **Document** - Generate docs from code (AI + human review)

````python
# Example: AI-Generated Code That Needs Human Review

def calculate_discount(price: float, discount_percent: float) -> float:
    """Calculate discounted price.
    
    AI generated this, but a human should verify:
    - Edge cases (negative values?)
    - Rounding behavior for currency
    - Discount limits (can it exceed 100%?)
    """
    return price * (1 - discount_percent / 100)

# Human-improved version with proper validation:
from decimal import Decimal, ROUND_HALF_UP

def calculate_discount_v2(
    price: Decimal, 
    discount_percent: Decimal,
    max_discount: Decimal = Decimal("50")
) -> Decimal:
    """Calculate discounted price with proper currency handling.
    
    Args:
        price: Original price (must be >= 0)
        discount_percent: Discount percentage (0-max_discount)
        max_discount: Maximum allowed discount (default 50%)
    
    Returns:
        Discounted price rounded to 2 decimal places
    
    Raises:
        ValueError: If inputs are out of valid range
    """
    if price < 0:
        raise ValueError("Price cannot be negative")
    if not 0 <= discount_percent <= max_discount:
        raise ValueError(f"Discount must be between 0 and {max_discount}%")
    
    discount_factor = 1 - discount_percent / 100
    result = price * discount_factor
    return result.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)
````

### 4.3 Code Review Checklist for AI-Generated Code

Always verify AI-generated code against this checklist:

- [ ] **Security** - No injection vulnerabilities, proper input validation
- [ ] **Edge Cases** - Handles null, empty, negative, overflow
- [ ] **Error Handling** - Appropriate exceptions, not silent failures
- [ ] **Performance** - Reasonable algorithmic complexity
- [ ] **Dependencies** - Uses existing project dependencies, not random packages
- [ ] **Style** - Matches project conventions
- [ ] **Tests** - Adequate coverage of happy path and edge cases
- [ ] **Documentation** - Accurate docstrings and comments

## 5. Code Review and Quality Assurance with AI

### 5.1 Using AI for Code Review

AI can assist with code review by analyzing:

- **Bugs and Logic Errors** - Pattern matching against common mistakes
- **Security Vulnerabilities** - OWASP Top 10, SQL injection, XSS
- **Performance Issues** - N+1 queries, unnecessary loops
- **Code Smells** - Long functions, deep nesting, code duplication
- **Best Practices** - Design patterns, SOLID principles

````python
# Example: Code with issues that AI should catch

def get_user_data(user_id):
    # SQL Injection vulnerability!
    query = f"SELECT * FROM users WHERE id = {user_id}"
    
    # No error handling
    result = db.execute(query)
    
    # Returning internal error details (information disclosure)
    return result

# AI-suggested improvements:
def get_user_data_secure(user_id: int) -> dict | None:
    """Fetch user data by ID with proper security.
    
    Args:
        user_id: The user's unique identifier
        
    Returns:
        User data dict or None if not found
        
    Raises:
        DatabaseError: If database operation fails
    """
    try:
        # Parameterized query prevents SQL injection
        query = "SELECT id, name, email FROM users WHERE id = ?"
        result = db.execute(query, (user_id,))
        return result.fetchone()
    except DatabaseError:
        logger.exception("Failed to fetch user %s", user_id)
        raise  # Re-raise without exposing internal details
````

## 6. Limitations and Considerations

### 6.1 Known Limitations of AI Coding Assistants

| Limitation | Description | Mitigation |
|------------|-------------|------------|
| **Hallucinations** | May invent non-existent APIs/functions | Always verify imports and APIs exist |
| **Outdated Knowledge** | Training data has a cutoff date | Provide current documentation |
| **Context Limits** | May lose track in long sessions | Break into smaller tasks |
| **Security Blindspots** | May miss subtle vulnerabilities | Use dedicated security tools |
| **Overconfidence** | Presents incorrect code confidently | Always review and test |
| **Pattern Matching** | May not understand true intent | Be explicit in requirements |

### 6.2 When NOT to Rely on AI

**Critical Areas Requiring Human Expertise:**

1. **Security-Critical Code** - Authentication, encryption, access control
2. **Financial Calculations** - Monetary transactions, tax calculations
3. **Legal/Compliance** - GDPR, HIPAA, industry regulations
4. **Architecture Decisions** - System design, technology choices
5. **Algorithmic Correctness** - When mathematical proof is needed
6. **Edge Cases in Production** - When failure has significant consequences

````python
# Example: AI might generate plausible-looking but incorrect code

# AI-generated (looks correct but has a subtle bug)
def calculate_compound_interest(principal, rate, years):
    return principal * (1 + rate) ** years

# The issue: rate should typically be expressed as decimal (0.05 for 5%)
# But some contexts expect percentage (5 for 5%)
# Without clear specification, AI might guess wrong

# Human-verified version with explicit handling:
def calculate_compound_interest_v2(
    principal: float,
    annual_rate_percent: float,  # Explicit: 5 means 5%
    years: int,
    compounds_per_year: int = 12
) -> float:
    """Calculate compound interest with explicit rate format.
    
    Args:
        principal: Initial amount
        annual_rate_percent: Annual interest rate as percentage (e.g., 5 for 5%)
        years: Number of years
        compounds_per_year: Compounding frequency (default: monthly)
    """
    rate_decimal = annual_rate_percent / 100
    n = compounds_per_year
    t = years
    return principal * (1 + rate_decimal / n) ** (n * t)

# Test both to verify
print(f"V1 (5% for 10 years): ${calculate_compound_interest(1000, 0.05, 10):.2f}")
print(f"V2 (5% for 10 years): ${calculate_compound_interest_v2(1000, 5, 10):.2f}")
````
